```python
def productivity_label(row):

    if row["idle_time"] > 60:
        return "idle"

    elif (
        row["keypresses"] > 5
        or row["typing_intensity"] > 5
    ):
        return "productive"

    elif row["mouse_clicks"] > 3:
        return "active"

    else:
        return "low_activity"
```

```python
df["label"] = df.apply(
    productivity_label,
    axis=1
)
```

```python
def format_idle_time(x):
    x = float(x)
    hours = int(x)
    minutes = int((x - hours) * 60)
    seconds = 0
    return f"{hours:02d}.{minutes:02d}.{seconds:02d}"

df["idle_time"] = df["idle_time"].apply(format_idle_time)
```
